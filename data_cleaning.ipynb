```python
import pandas as pd

df = pd.read_csv("AirfoilDatset.csv")
df.shape
df.head()
df.columns
df.isnull().sum()
df.describe()
```

```python
len(df)
```

```python
df_clean = df.copy()

for target in ["coefficientLift", "coefficientDrag"]:
    Q1 = df_clean[target].quantile(0.25)
    Q3 = df_clean[target].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df_clean = df_clean[(df_clean[target] >= lower) & (df_clean[target] <= upper)]

len_before = len(df)
len_after = len(df_clean)
removed = len_before - len_after
percent_removed = removed / len_before * 100

len_before, len_after, removed, percent_removed
```

```python
df_clean.to_csv("AirfoilDatset_cleaned.csv", index=False)
```
